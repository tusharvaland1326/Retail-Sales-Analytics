```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Imported Successfully!")
```

```python
df = pd.read_csv("SampleSuperstore.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
print(df.shape)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    df.corr(numeric_only=True),
    annot=True,
    cmap="Blues"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

category_profit.plot(kind="bar", color="green")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

segment_sales.plot(kind="bar", color="purple")

plt.title("Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Sales")

plt.show()
```

```python
segment_profit = df.groupby("Segment")["Profit"].sum().sort_values(ascending=False)

plt.figure(figsize=(8,5))

segment_profit.plot(kind="bar", color="teal")

plt.title("Total Profit by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Profit")

plt.show()
```

```python
state_sales = df.groupby("State")["Sales"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

state_sales.plot(kind="bar", color="royalblue")

plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

```python
city_profit = df.groupby("City")["Profit"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))

city_profit.plot(kind="bar", color="green")

plt.title("Top 10 Cities by Profit")
plt.xlabel("City")
plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(7,7))

df["Ship Mode"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Ship Mode Distribution")
plt.ylabel("")

plt.show()
```

```python
plt.figure(figsize=(7,5))

sns.countplot(
    x="Category",
    data=df,
    order=df["Category"].value_counts().index
)

plt.title("Product Category Distribution")
plt.xlabel("Category")
plt.ylabel("Number of Orders")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Discount"], bins=20)

plt.title("Discount Distribution")
plt.xlabel("Discount")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Sales"], bins=30)

plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.hist(df["Profit"], bins=30)

plt.title("Profit Distribution")
plt.xlabel("Profit")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.scatter(df["Sales"], df["Profit"])

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
subcategory_sales = (
    df.groupby("Sub-Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(12,6))

subcategory_sales.plot(kind="bar")

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)

plt.show()
```
